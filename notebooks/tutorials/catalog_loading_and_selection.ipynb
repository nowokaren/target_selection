# Catalog loading and target selection

This notebook demonstrates the new source configuration and the public target-loading API. It is independent of the historical LaStBeRu and QSO notebooks.

```python
from target_selection import target_io

print(target_io.list_sources())
print(target_io.list_adapters())
print(target_io.describe_source("qso_extended"))
```

## Load a configured source

The source configuration defines the file, input columns, category parser, and capabilities.

```python
catalog = target_io.load("qso_extended", filters={"category": [1]})
catalog.head()
```

## Load a local table with explicit column definitions

```python
import pandas as pd
local_rows = pd.DataFrame({
    "object_id": ["example-1", "example-2"],
    "ra": [265.0, 266.0],
    "dec": [-29.0, -28.0],
    "status": [1, 2],
})
local_catalog = target_io.load(
    local_rows, source_name="example_catalog",
    id_column="object_id", ra_column="ra", dec_column="dec",
    category_column="status", category_parser="identity",
    filters={"category": [1]},
)
local_catalog
```

## Add optional data

Use add_data only when extra information is needed after loading. It does not silently query an external service.

```python
# Synthetic optional data illustrates an explicit enrichment step.
extra = pd.DataFrame({"target_id": local_catalog["target_id"], "magnitude_i": 18.5})
enriched = target_io.add_data(local_catalog, extra, provider="example_provider", fields=["magnitude_i"])
enriched
```
