# 2. Seasonal visibility selection

This notebook is the template for selecting targets before an HSH or JS observing season.

```python
from target_selection import target_io
from target_selection.tasks.visibility import evaluate_visibility, save_visibility_plots, make_visibility_sequence

SOURCE = "qso_extended"
OBSERVATORY = "CASLEO"
START_DATE = "2026-10-01"
END_DATE = "2026-12-31"
MIN_ALTITUDE_DEG = 30.0
MIN_VISIBLE_MINUTES = 90.0
```

```python
targets = target_io.load(SOURCE, filters={})
visibility = evaluate_visibility(targets, start_date=START_DATE, end_date=END_DATE, observatory=OBSERVATORY, minimum_altitude_deg=MIN_ALTITUDE_DEG, minimum_observable_minutes=MIN_VISIBLE_MINUTES, output_path="outputs/season/visibility.csv")
selected = visibility.loc[visibility["selected_for_visibility"].astype(bool)].copy()
```

```python
save_visibility_plots(targets, start_date=START_DATE, end_date=END_DATE, output_dir="outputs/season/visibility_plots", observatory=OBSERVATORY, minimum_altitude_deg=MIN_ALTITUDE_DEG, minimum_observable_minutes=MIN_VISIBLE_MINUTES, selection=visibility)
make_visibility_sequence(visibility, "outputs/season/visibility_sequence.pdf", observatory=OBSERVATORY)
```

The resulting selected table is the input to survey-specific planning for HSH or JS. The future provider-specific steps should add observation history and survey photometry without changing this visibility interface.
